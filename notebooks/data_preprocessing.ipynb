```python
import pandas as pd
import re
from urllib.parse import urlparse

def extract_features(url):

    parsed = urlparse(url)
    domain = parsed.netloc

    features = {}

    features['URLLength'] = len(url)
    features['DomainLength'] = len(domain)

    features['IsDomainIP'] = 1 if re.match(r'\d+\.\d+\.\d+\.\d+', domain) else 0

    features['TLDLength'] = len(domain.split('.')[-1]) if '.' in domain else 0

    features['NoOfSubDomain'] = max(domain.count('.') - 1, 0)

    features['HasObfuscation'] = 1 if '%' in url else 0

    features['NoOfObfuscatedChar'] = url.count('%')

    features['NoOfLettersInURL'] = sum(c.isalpha() for c in url)

    features['NoOfDegitsInURL'] = sum(c.isdigit() for c in url)

    features['NoOfEqualsInURL'] = url.count('=')

    features['NoOfQMarkInURL'] = url.count('?')

    features['NoOfAmpersandInURL'] = url.count('&')

    features['NoOfOtherSpecialCharsInURL'] = len(re.findall(r'[^\w]', url))

    features['IsHTTPS'] = 1 if parsed.scheme == "https" else 0

    features['Bank'] = 1 if "bank" in url.lower() else 0

    features['Pay'] = 1 if "pay" in url.lower() else 0

    features['Crypto'] = 1 if "crypto" in url.lower() else 0

    return features


df_features = df['url'].apply(extract_features).apply(pd.Series)

df_final = pd.concat([df_features, df['label']], axis=1)

df_final.to_csv("../notebooks/cleaned_phishing.csv", index=False)

print("Preprocessing completed")
```

```python
import re 
def extract_url_features(url):
    return pd.Series({
        'url_length': len(url),
        'dot_count':url.count('.'),
        'hyphen_count':url.count('-'),
        'slash_count':url.count('/'),
        'digit_count': sum(char.isdigit() for char in url),
        'has_at_symbol': 1 if '@' in url else 0,
        'has_ip': 1 if re.match(r'\d+\.\d+\.\d+\.\d+', url) else 0
    })
url_features = df['URL'].apply(extract_url_features)
display(url_features.head())
```
